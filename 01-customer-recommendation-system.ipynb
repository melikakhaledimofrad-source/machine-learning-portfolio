  #  Smart Pet Store Recommendation System
  A simple Machine Learning model using **K-Nearest Neighbors (KNN)** and Euclidean distance to recommend pet products based on customer behavioral similarity.


```python
import pandas as pd
from sklearn.neighbors import NearestNeighbors
```

```python
K_Neighbors=4
orders = [
    {"customer_id": 1, "products": ["dry_food", "treat", "treat"]},
    {"customer_id": 2, "products": ["treat", "toy"]},
    {"customer_id": 3, "products": ["dry_food", "dry_food", "treat", "toy"]},
    {"customer_id": 4, "products": ["toy", "toy", "toy"]}
]
```

```python
new_orders = ["dry_food", "treat"]
```

```python
orders.append({"customer_id":5, "products": new_orders })
```

```python
df = pd.DataFrame(orders)
df
```

```python
df = df.explode("products")
```

```python
result_orders= pd.crosstab(df['customer_id'], df['products'])
result_orders
```

```python
old_df = result_orders.iloc[:-1].values
new_df = result_orders.iloc[-1].values.reshape(1,-1)
old_df
```

```python
new_df
```

  ###  Finding Nearest Neighbors & Recommendations
  Using Scikit-Learn's `NearestNeighbors` to find the most similar existing customer and suggest missing products.


```python
K_nearest = NearestNeighbors(n_neighbors = K_Neighbors )
K_nearest.fit(old_df)
Distance , indexes = K_nearest.kneighbors(new_df)
Distance , indexes
```

```python
sorted_by_distance = result_orders.iloc[indexes[0]].values
sorted_by_distance
```

```python
no_buy = sorted_by_distance[0]- new_df
df_no_buy = pd.DataFrame(no_buy, columns=result_orders.columns)
```

```python
recommended = []
for i in df_no_buy.columns:
    if df_no_buy[i].values>0:
        recommended.append(i)
        print("----------------------Recommend System--------------------")
        print(f">>>>>>>>>>>>>>>>>>>> Customer ID:{orders[-1]['customer_id']} <<<<<<<<<<<<<<<<<<<<<<<")
        print(f"----------------------------------------------------------")
        print(f"|                         {recommended[0]}                          |")
        print(f"----------------------------------------------------------")
```

---

##  Conclusion & Results
By applying the **K-Nearest Neighbors (KNN)** algorithm and calculating Euclidean distances, we successfully identified the customer with the closest purchasing behavior to our new user (Customer ID: 5). 

- **Nearest Neighbor Match:** Customer at index `0` (Distance: `1.0`)
- **Recommended Item:** **`treat`**

This lightweight recommendation engine demonstrates how basic mathematical distance metrics can be leveraged to build intelligent, data-driven product recommendation systems for e-commerce and retail platforms.
