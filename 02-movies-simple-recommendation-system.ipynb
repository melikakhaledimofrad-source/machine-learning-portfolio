```python
import pandas as pd
import numpy as np 
from sklearn.neighbors import NearestNeighbors
```

```python
watch_history = [
    {"user_id": 1, "genres": ["sci-fi", "adventure"]},
    {"user_id": 1, "genres": ["comedy", "animation"]},

    {"user_id": 2, "genres": ["drama", "romance"]},
    {"user_id": 2, "genres": ["romance", "comedy"]},

    {"user_id": 3, "genres": ["sci-fi", "action"]},
    {"user_id": 3, "genres": ["action", "adventure"]},

    {"user_id": 4, "genres": ["documentary", "drama"]},
    {"user_id": 4, "genres": ["history", "documentary"]},

    {"user_id": 5, "genres": ["animation", "comedy"]},
    {"user_id": 5, "genres": ["adventure", "animation"]},

    {"user_id": 6, "genres": ["sci-fi", "thriller"]},
    {"user_id": 6, "genres": ["mystery", "thriller"]}
]
```

```python
new_user = {"user_id": 7, "genres": ["mystery", "thriller", "adventure"]}
```

```python
watch_history.append(new_user)
```

```python
df = pd.DataFrame(watch_history)
df
```

```python
df = df.explode('genres')
df
```

```python
df_result = pd.crosstab(df['user_id'],df['genres'])
df_result
```

```python
X_train = df_result.iloc[:-1].values
X_new_user = df_result.iloc[-1:].values
X_new_user
```

```python
df_test = pd.DataFrame(X_train)
df_test
```

```python
model = NearestNeighbors(n_neighbors=5).fit(X_train)
model
Distance , indexes = model.kneighbors(X_new_user)
```

```python
print("Distances:",Distance[0])
print("Indexes:", indexes[0])
```

```python
best_match_idx = indexes[0][0] 
recommender = df_result.iloc[best_match_idx]
similar_user_id = df_result.index[best_match_idx]
print(similar_user_id)
```

```python
print(f"New user saw: {new_user['genres']}")
print(f"Similar User ID is ---> {similar_user_id} (at index {best_match_idx})")
```

```python
similar_genres = recommender[recommender > 0].index.values
print(f"Similar user saw genres: {similar_genres}")
print(f"Similar user :{similar_genres}\tVS\t New user:{new_user['genres']}")
```

```python
list_recommender = [genre for genre in similar_genres if genre not in new_user["genres"]]
print("----------------------------------------")
print("<<<Recommendation:", list_recommender,">>>")
print("----------------------------------------")
```
